## 4. The unit of analysis, as a real dataframe

**Unit of analysis:** One row represents one anonymized content page.

**Features:** Content age, time since last update, impressions, average search position, and CTR are candidate observable signals.

**Target sketch:** `is_declining_proxy` is 1 when `trend_direction` equals `down`, and 0 otherwise. This is a starter proxy, not a future outcome.

**Limitation:** Pages belonging to the same client may share characteristics. Evaluation should therefore hold out clients rather than randomly splitting individual rows, to reduce the risk of overly optimistic results.


```python

from pathlib import Path
import subprocess
import pandas as pd

# Find the starter dataset, cloning your public repository if needed.
data_path = Path("data/raw/content_refresh_anonymized.csv")

if not data_path.exists():
    repo_dir = Path("/content/ml-internship")

    if not repo_dir.exists():
        subprocess.run(
            [
                "git", "clone", "--depth", "1",
                "https://github.com/Dushyant1607/ml-internship.git",
                str(repo_dir)
            ],
            check=True
        )

    data_path = repo_dir / "data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

# One row represents one anonymized content page.
df["is_declining_proxy"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

columns = [
    "content_id",
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "avg_position",
    "ctr",
    "trend_direction",
    "is_declining_proxy",
]

display(df[columns].head(10))

print(f"Rows (content pages): {len(df):,}")
print(f"Columns: {len(df.columns)}")
print(
    "Declining proxy rate:",
    f"{df['is_declining_proxy'].mean():.1%}"
)

```

## 5. Why ML beats a fixed rule here

A fixed rule such as prioritizing pages that have not been updated for 180 days and have at least 500 impressions is easy to explain and provides a useful baseline. However, it relies on thresholds chosen in advance and may miss interactions between content age, impressions, CTR, and search position.

A model can learn combinations of these signals and potentially improve the ranking. I will only claim that ML adds value if it improves an appropriate evaluation metric on held-out data compared with the baseline. If the fixed rule performs as well or better, its simplicity and interpretability may make it the better choice. The output supports human review; it does not automatically authorize content changes.
