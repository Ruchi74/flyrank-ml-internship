## 2. The question: decision, action, cost of a wrong call

**Research question:** Among the available content pages, which should an SEO or editorial reviewer inspect first for a possible refresh, expansion, protection, pruning, or monitoring?

**Unit of analysis:** One anonymized content page, represented by one row in the starter dataset.

**Decision and action:** The output will be a ranked list of review candidates with a priority score and understandable reason codes. A reviewer will inspect the evidence and context before choosing an action.

**Cost of a wrong recommendation:** A false positive could waste review time or lead to an unnecessary content edit. A false negative could cause a genuine opportunity or declining page to be overlooked. The ranking should therefore support human judgment rather than automatically trigger changes.

**Why data or ML might help:** Page age, update recency, impressions, search position, CTR, and content depth may interact in ways that a single hand-written rule misses. I will compare a transparent baseline with a readable model and retain the model only if it adds useful value.

## 3. Quick look at the data (2-3 real numbers)

I will use the 30,000-row anonymized starter dataset to check whether there is a practical opportunity for content review. I will examine the share of pages currently labelled as declining and count pages that are both stale and still visible in search. These measurements can establish whether the lane is worth investigating, but they do not show that refreshing a page will cause it to recover.


```python

from pathlib import Path
import subprocess
import pandas as pd

relative_path = Path("data/raw/content_refresh_anonymized.csv")

# Look for the dataset in common Colab locations.
candidates = [
    relative_path,
    Path("/content/ml-internship") / relative_path,
    Path("/content/flyrank-ml-internship-starter") / relative_path,
]

data_path = next((p for p in candidates if p.exists()), None)

# If needed, clone your public repository to access the starter CSV.
if data_path is None:
    repo_dir = Path("/content/ml-internship")
    if not repo_dir.exists():
        subprocess.run(
            [
                "git", "clone", "--depth", "1",
                "https://github.com/Dushyant1607/ml-internship.git",
                str(repo_dir)
            ],
            check=True
        )
    data_path = repo_dir / relative_path

df = pd.read_csv(data_path)

required = {
    "trend_direction",
    "days_since_last_update",
    "impressions_90d",
}
missing = required - set(df.columns)
if missing:
    raise ValueError(f"Missing required columns: {sorted(missing)}")

declining = df["trend_direction"].str.lower().eq("down")
stale_visible = (
    (df["days_since_last_update"] >= 180)
    & (df["impressions_90d"] >= 500)
)

print(f"Dataset size: {len(df):,} rows × {len(df.columns)} columns")
print(
    f"Pages labelled declining: {declining.sum():,} "
    f"({declining.mean():.1%})"
)
print(
    f"Stale and visible review candidates: "
    f"{stale_visible.sum():,} ({stale_visible.mean():.1%})"
)

```
